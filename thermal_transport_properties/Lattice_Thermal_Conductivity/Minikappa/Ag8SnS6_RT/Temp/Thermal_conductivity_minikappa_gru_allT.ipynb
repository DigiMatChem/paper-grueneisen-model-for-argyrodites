```python
import pandas as pd 
import os
```

```python
df = pd.DataFrame(index= list(range(0,950,50)), columns= ['x','y','z'])
```

```python
file_names = [i for i in os.listdir() if i.startswith('minikappa')]
```

```python
for file in file_names:
    temp = int(file.split('-')[1])
    with open(file, 'r') as f:
        data = f.readlines()[-1].strip('\n').split('  ')
    df.loc[temp,'x'] = float(data[0]) 
    df.loc[temp,'y'] = float(data[4]) 
    df.loc[temp,'z'] = float(data[8])
```
