```python
import pandas as pd
from mp_api.client import MPRester, MPRestError
from emmet.core.summary import HasProps
from empiricalthermalconductivity import thermcond
from pymatgen.analysis.elasticity import ElasticTensor
from pymatgen.core import Species, Structure
import warnings
warnings.filterwarnings('ignore')
```

```python
#Assign your values here for elastic constants in GPa
young_mod = 29.38
bulk_mod = 37.86
shear_mod = 11.32

# load your structure and get n_atoms_list and atomic_weight list
struct = Structure.from_file('POSCAR')
natoms_list= []
atomic_weight_list=[]
vol = struct.volume
for k, v in struct.composition.get_el_amt_dict().items():
    sp = Species(k).atomic_mass
    natoms_list.append(v)
    atomic_weight_list.append(sp)
    
kappa_cahill = thermcond.thermal_cond_cahill(natoms_list=natoms_list, atomic_weight_list=atomic_weight_list,
                     vol=vol, G=shear_mod, K=bulk_mod)
kappa_agne = thermcond.thermal_cond_diffuson_agne(natoms_list=natoms_list, atomic_weight_list=atomic_weight_list,
                     vol=vol, G=shear_mod, K=bulk_mod)
kappa_clarke = thermcond.thermal_cond_clarke(natoms_list=natoms_list, atomic_weight_list=atomic_weight_list,
                     vol=vol, E=young_mod)
kappa_slack = thermcond.thermal_cond_slack(natoms_list=natoms_list, atomic_weight_list=atomic_weight_list,
                     vol=vol, G=shear_mod, K=bulk_mod, t=600)
kappa_mixed = thermcond.thermal_cond_latt_mixed(natoms_list=natoms_list, atomic_weight_list=atomic_weight_list,
                     vol=vol, K=bulk_mod, E=young_mod, G=shear_mod, t=600)

# sound velocities
v_l, v_t, v_a, v_s = thermcond.sound_velocities(atomic_weight_list=atomic_weight_list, natoms_list=natoms_list, vol=vol,
                                                        E=young_mod,G=shear_mod, K=bulk_mod)

#Gruneisen Param
Gru= thermcond.Gruneisen_param(v_l=v_l, v_t=v_t)
```
