Checking for significant COBI interactions to decide which interactions to consinder for the multi-center-COBIs.

I added the translation vector and the bond length to the objects, too, in an attempt to get the bond angle amd the distance between the terminating atoms automatically, but in the end I had to do it manually via VESTA, because of how different the translation vector is handled in LOBSTER for the ICOBI and N-ICOBI calculations and VESTA. There was a mismatch that I could not fugure out how to solve, so in the end I did it manually via VESTA.

```python
sig_icobi_RT = [] # significant ICOBI values
sig_icobi_LT = []
sig_icobi_Ge = []
sig_icobi_Si = []

for i,j in enumerate(icobi_RT_dict['list_icohp']):
    sumSpin_RT = sum(list(icobi_RT_dict['list_icohp'][i].values()))
    if sumSpin_RT > 0.25: 
        sig_icobi_RT.append([icobi_RT_dict['list_atom2'][i],icobi_RT_dict['list_atom1'][i], icobi_RT_dict['list_translation'][i], icobi_RT_dict['list_length'][i],f'{sumSpin_RT:.2f}'])
for i,j in enumerate(icobi_LT_dict['list_icohp']):
    sumSpin_LT = sum(list(icobi_LT_dict['list_icohp'][i].values()))
    if sumSpin_LT > 0.25: 
        sig_icobi_LT.append([icobi_LT_dict['list_atom2'][i],icobi_LT_dict['list_atom1'][i], icobi_LT_dict['list_translation'][i], icobi_LT_dict['list_length'][i],f'{sumSpin_LT:.2f}'])
for i,j in enumerate(icobi_Ge_dict['list_icohp']):
    sumSpin_Ge = sum(list(icobi_Ge_dict['list_icohp'][i].values()))
    if sumSpin_Ge > 0.25: 
        sig_icobi_Ge.append([icobi_Ge_dict['list_atom2'][i],icobi_Ge_dict['list_atom1'][i], icobi_Ge_dict['list_translation'][i], icobi_Ge_dict['list_length'][i],f'{sumSpin_Ge:.2f}'])
for i,j in enumerate(icobi_Si_dict['list_icohp']):
    sumSpin_Si = sum(list(icobi_Si_dict['list_icohp'][i].values()))
    if sumSpin_Si > 0.25: 
        sig_icobi_Si.append([icobi_Si_dict['list_atom2'][i],icobi_Si_dict['list_atom1'][i], icobi_Si_dict['list_translation'][i], icobi_Si_dict['list_length'][i],f'{sumSpin_Si:.2f}'])

print("RT \n ", sig_icobi_RT, "\n LT \n", sig_icobi_LT, "\n Ge \n", sig_icobi_Ge, "\n Si \n", sig_icobi_Si)
```
